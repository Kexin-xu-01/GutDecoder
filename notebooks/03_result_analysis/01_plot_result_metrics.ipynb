```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

import matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display

def plot_summary_bar(best_df: pd.DataFrame, outdir: str | None = None, show: bool = False):
    """
    Barplot of highest_pearson_mean ± highest_pearson_std per run.
    - Grouped and colored by dataset
    - Within each dataset, runs sorted by mean desc
    - Each bar labeled with num_genes, offset to avoid overlapping error bar

    Args:
        best_df (pd.DataFrame): from summarize_runs
        outdir (str|None): directory to save plot
        show (bool): if True, display inline in Jupyter

    Returns:
        matplotlib.figure.Figure
    """
    if best_df.empty:
        print("[plot_summary_bar] Empty DataFrame.")
        return None

    # Ensure proper types
    df = best_df.copy()
    df["highest_pearson_mean"] = pd.to_numeric(df["highest_pearson_mean"], errors="coerce")
    df["highest_pearson_std"] = pd.to_numeric(df["highest_pearson_std"], errors="coerce").fillna(0.0)
    df["num_genes"] = pd.to_numeric(df["num_genes"], errors="coerce").fillna(0).astype(int)
    df["dataset"] = df["dataset"].astype(str).fillna("Unknown")

    # Sort datasets and entries
    datasets = sorted(df["dataset"].unique().tolist())
    palette = plt.get_cmap("tab10", len(datasets))
    color_map = {ds: palette(i) for i, ds in enumerate(datasets)}

    # Order: dataset then mean desc
    ordered = []
    for ds in datasets:
        sub = df[df["dataset"] == ds].sort_values("highest_pearson_mean", ascending=False)
        ordered.append(sub)
    df = pd.concat(ordered)

    # Build x-axis
    labels = (df["dataset"] + " | " + df["gene_list"]).tolist()
    x = np.arange(len(df))
    means = df["highest_pearson_mean"].to_numpy()
    errs = df["highest_pearson_std"].to_numpy()
    colors = [color_map[ds] for ds in df["dataset"]]
    nums = df["num_genes"].to_numpy()

    # Plot
    fig, ax = plt.subplots(figsize=(max(12, len(df)*0.5), 8))
    bars = ax.bar(x, means, yerr=errs, color=colors, capsize=4, edgecolor="black")

    # Annotate number of genes offset away from error bars
    for i, (bar, val, err, ng) in enumerate(zip(bars, means, errs, nums)):
        height = bar.get_height()
        # offset direction: above positive bar, below negative
        offset = 0.01
        va = "bottom" if height >= 0 else "top"
        ax.annotate(str(ng),
                    xy=(bar.get_x() + bar.get_width()/2, height + (err if height >=0 else -err)),
                    xytext=(0, 5 if height >= 0 else -5), textcoords="offset points",
                    ha="center", va=va, fontsize=9, rotation=0)

    # Styling
    ax.set_xticks(x)
    ax.set_xticklabels(labels, rotation=60, ha="right")
    ax.set_ylabel("Pearson correlation (mean ± std)")
    ax.set_title("Run summary grouped by dataset (label = num_genes)")
    ax.grid(True, axis="y", linestyle=":", alpha=0.4)

    # Legend
    handles = [plt.Rectangle((0,0),1,1, color=color_map[ds], edgecolor="black") for ds in datasets]
    ax.legend(handles, datasets, title="Dataset", bbox_to_anchor=(1.02, 1), loc="upper left")

    fig.tight_layout()

    # Save / show
    if outdir:
        outdir = Path(outdir); outdir.mkdir(parents=True, exist_ok=True)
        p = outdir / "summary_barplot.png"
        fig.savefig(p, dpi=200, bbox_inches="tight")
    if show:
        display(fig)

    return fig



def plot_summary_genes_vs_mean(best_df: pd.DataFrame, outdir: str | None = None, show: bool = False):
    """
    Scatter: x = highest_pearson_mean, y = num_genes, colored by dataset.
    Saves to outdir/'summary_genes_vs_mean.png' if outdir is provided.

    Returns:
        matplotlib.figure.Figure
    """
    if best_df is None or best_df.empty:
        print("[plot_summary_genes_vs_mean] Empty DataFrame.")
        return None

    df = best_df.copy()

    need = {"dataset", "num_genes", "highest_pearson_mean"}
    missing = need - set(df.columns)
    if missing:
        raise ValueError(f"[plot_summary_genes_vs_mean] Missing columns: {missing}")

    df["dataset"] = df["dataset"].astype("string").fillna("Unknown")
    df["num_genes"] = pd.to_numeric(df["num_genes"], errors="coerce")
    df["highest_pearson_mean"] = pd.to_numeric(df["highest_pearson_mean"], errors="coerce")

    # Drop rows where either axis is NaN
    df = df.dropna(subset=["num_genes", "highest_pearson_mean"])
    if df.empty:
        print("[plot_summary_genes_vs_mean] No rows with both num_genes and pearson mean.")
        return None

    # Color map per dataset
    datasets = sorted(df["dataset"].dropna().unique().tolist())
    cmap = plt.get_cmap("tab10", len(datasets))
    color_map = {ds: cmap(i) for i, ds in enumerate(datasets)}

    fig, ax = plt.subplots(figsize=(8, 6))
    for ds in datasets:
        sub = df[df["dataset"] == ds]
        ax.scatter(
            sub["highest_pearson_mean"].to_numpy(dtype=float),
            sub["num_genes"].to_numpy(dtype=float),
            label=str(ds),
            alpha=0.85,
            edgecolors="black",
            linewidths=0.5,
        )

    ax.set_xlabel("Mean Pearson correlation")
    ax.set_ylabel("Number of genes")
    ax.set_title("Gene count vs performance (colored by dataset)")
    ax.grid(True, linestyle=":", alpha=0.4)
    ax.legend(title="Dataset", bbox_to_anchor=(1.02, 1), loc="upper left")
    fig.tight_layout()

    if outdir:
        outdir = Path(outdir); outdir.mkdir(parents=True, exist_ok=True)
        p = outdir / "summary_genes_vs_mean.png"
        fig.savefig(p, dpi=200, bbox_inches="tight")

    if show:
        display(fig)

    return fig


from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from adjustText import adjust_text

def plot_summary_patches_vs_mean(
    best_df: pd.DataFrame,
    outdir: str | None = None,
    show: bool = False,
    figsize=(9, 7),
    size_scale: float = 0.05,   # scaling factor for dot size from num_genes
    min_size: float = 30,       # minimum dot size for visibility
    size_legend_values: list[int] = None,  # values to show in size legend
):
    """
    Scatter: 
      x = highest_pearson_mean
      y = num_training_patches_mean
      dot size = num_genes
      color = dataset
      label = gene_list (without .json), repelled with connectors
      size legend shows num_genes examples

    Args:
        best_df (pd.DataFrame): summary dataframe from summarize_runs
        outdir (str|None): directory to save the figure
        show (bool): if True, display in Jupyter
        figsize (tuple): figure size
        size_scale (float): scaling factor for num_genes -> marker size
        min_size (float): minimum marker size
        size_legend_values (list[int]): values of num_genes to display in size legend

    Returns:
        matplotlib.figure.Figure | None
    """
    if best_df is None or best_df.empty:
        print("[plot_summary_patches_vs_mean] Empty DataFrame.")
        return None

    need = {"dataset", "gene_list", "highest_pearson_mean", "num_training_patches_mean", "num_genes"}
    missing = need - set(best_df.columns)
    if missing:
        raise ValueError(f"[plot_summary_patches_vs_mean] Missing columns: {missing}")

    df = best_df.copy()

    # Clean and coerce types
    df["dataset"] = df["dataset"].astype(str).fillna("Unknown")
    df["gene_list"] = df["gene_list"].astype(str).fillna("Unknown").str.replace(".json", "", regex=False)
    df["highest_pearson_mean"] = pd.to_numeric(df["highest_pearson_mean"], errors="coerce")
    df["num_training_patches_mean"] = pd.to_numeric(df["num_training_patches_mean"], errors="coerce")
    df["num_genes"] = pd.to_numeric(df["num_genes"], errors="coerce")

    # Drop rows with missing values
    df = df.dropna(subset=["highest_pearson_mean", "num_training_patches_mean", "num_genes"])
    if df.empty:
        print("[plot_summary_patches_vs_mean] No valid rows to plot.")
        return None

    # Map datasets to colors
    datasets = sorted(df["dataset"].unique().tolist())
    cmap = plt.get_cmap("tab10", len(datasets))
    color_map = {ds: cmap(i) for i, ds in enumerate(datasets)}

    fig, ax = plt.subplots(figsize=figsize)
    texts = []

    for ds in datasets:
        sub = df[df["dataset"] == ds]
        sc = ax.scatter(
            sub["highest_pearson_mean"],
            sub["num_training_patches_mean"],
            s=(sub["num_genes"] * size_scale).clip(lower=min_size),
            alpha=0.9,
            edgecolors="black",
            linewidths=0.5,
            color=color_map[ds],
            label=ds,
        )
        for _, row in sub.iterrows():
            texts.append(
                ax.text(
                    row["highest_pearson_mean"],
                    row["num_training_patches_mean"],
                    row["gene_list"],
                    fontsize=8,
                )
            )

    # Repel labels with connectors
    adjust_text(
        texts,
        ax=ax,
        expand_points=(1.2, 1.4),
        expand_text=(1.2, 1.4),
        arrowprops=dict(arrowstyle="-", color="gray", lw=0.5)
    )

    # Axis labels & styling
    ax.set_xlabel("Mean Pearson correlation")
    ax.set_ylabel("Mean number of training patches per split")
    ax.set_title("Training patches vs performance (dot size = num_genes)")
    ax.grid(True, linestyle=":", alpha=0.4)

    # Legend by dataset
    handles, labels = ax.get_legend_handles_labels()
    leg1 = ax.legend(handles, labels, title="Dataset", bbox_to_anchor=(1.02, 1), loc="upper left")

    # Add size legend
    if size_legend_values is None:
        # pick sensible defaults based on data
        size_legend_values = [int(df["num_genes"].min()), 
                              int(df["num_genes"].median()), 
                              int(df["num_genes"].max())]
        size_legend_values = sorted(set(size_legend_values))  # remove duplicates

    size_handles = [
        plt.scatter([], [], 
                    s=max(min_size, val*size_scale), 
                    color="gray", alpha=0.6, edgecolors="black")
        for val in size_legend_values
    ]
    leg2 = ax.legend(size_handles, [f"{v} genes" for v in size_legend_values],
                     title="Num genes (dot size)",
                     bbox_to_anchor=(1.02, 0.4), loc="upper left")
    ax.add_artist(leg1)

    fig.tight_layout()

    if outdir:
        outdir = Path(outdir); outdir.mkdir(parents=True, exist_ok=True)
        p = outdir / "summary_patches_vs_mean.png"
        fig.savefig(p, dpi=200, bbox_inches="tight")

    if show:
        display(fig)

    return fig
```

```python
DEFAULT_RUNS_ROOT = "/project/simmons_hts/kxu/hest/eval/ST_pred_results"
DEFAULT_SPLITS_ROOT = "/project/simmons_hts/kxu/hest/eval/data"
DEFAULT_CURATED_XLSX = "/project/simmons_hts/kxu/hest/curated_gene_list.xlsx"
DEFAULT_EXTRA_METADATA = "/project/simmons_hts/kxu/hest/hest_directory.csv"

def generate_all_plots(run: str,
                       runs_root=DEFAULT_RUNS_ROOT,
                       splits_root=DEFAULT_SPLITS_ROOT,
                       curated_xlsx: str | None = DEFAULT_CURATED_XLSX,
                       extra_metadata_csv: str | None = DEFAULT_EXTRA_METADATA,
                       top_n=30,
                       show=False) -> dict:
    """
    New workflow:
      - Always produce base plots (gene_barplot, gene_hist, per_sample).
      - Only produce grouped plots if the curated list and/or metadata are provided & present.
    Saves into <runs_root>/<run>/plots with simple filenames.
    """
    # Where to save
    run_dir = Path(runs_root) / run
    outdir = run_dir / "plots"
    outdir.mkdir(parents=True, exist_ok=True)

    # 1) Best model + gene correlations (k-fold)
    best, dataset_name, df_genes = extract_best_model_gene_corrs(run, runs_root=runs_root, verbose=False)
    encoder_name = best.get("encoder_name", "Unknown")

    # 2) Splits (+ optional HEST directory metadata inside get_test_splits)
    #    Note: get_test_splits already merges extra metadata for XeniumPR1/pilot if CSV exists.
    df_splits = get_test_splits(run, runs_root=runs_root,
                                splits_root=splits_root,
                                extra_metadata_csv=extra_metadata_csv)

    # 3) Base per-split long df
    df_long = merge_kfold_gene_corrs_with_test_metadata(df_genes, df_splits)

    arts = {}

    # ---------------- Base plots (always) ----------------
    # 3a) Gene barplot (ungrouped)
    fig = plot_gene_correlation_barplot(
        df_genes,dataset_name, encoder_name
    )
    p = outdir / "gene_barplot.png"
    fig.savefig(p, dpi=200, bbox_inches="tight"); plt.close(fig)
    arts["gene_barplot"] = p

    # 3b) Histogram
    fig = plot_gene_correlation_histogram(df_genes, dataset_name, encoder_name)
    p = outdir / "gene_hist.png"
    fig.savefig(p, dpi=200, bbox_inches="tight"); plt.close(fig)
    arts["gene_hist"] = p

    # 3c) Per-sample (ungrouped if no group col present)
    fig = plot_corrs_by_sample(df_long, dataset_name, encoder_name, group_by="Sample_type")
    p = outdir / "per_sample.png"
    fig.savefig(p, dpi=200, bbox_inches="tight"); plt.close(fig)
    arts["per_sample"] = p

    # ---------------- Grouped plots (only if provided) ----------------
    # Curated list → annotate genes and per-sample df, then group by panel/cell_type/condition
    curated_ok = (isinstance(curated_xlsx, str) and Path(curated_xlsx).exists())
    if curated_ok:
        df_genes_annot = annotate_genes_with_curated(df_genes, curated_xlsx)
        df_long_annot  = annotate_genes_with_curated(df_long,  curated_xlsx)
        display(df_long_annot)

        # Gene grouped barplots
        for group_col in ["panel", "cell_type", "condition"]:
            if group_col in df_genes_annot and df_genes_annot[group_col].notna().any():
                fig = plot_gene_correlation_barplot_grouped(
                    df_genes_annot, 
                    group_by=group_col
                )
                p = outdir / f"gene_barplot_by_{group_col}.png"
                fig.savefig(p, dpi=200, bbox_inches="tight"); plt.close(fig)
                arts[f"gene_barplot_by_{group_col}"] = p

        # Per-sample grouped by curated columns
        for group_col in ["panel", "cell_type", "condition"]:
            if group_col in df_long_annot and df_long_annot[group_col].notna().any():
                fig = plot_corrs_by_sample(df_long_annot, dataset_name, encoder_name,
                                           group_by=group_col)
                p = outdir / f"per_sample_by_{group_col}.png"
                fig.savefig(p, dpi=200, bbox_inches="tight"); plt.close(fig)
                arts[f"per_sample_by_{group_col}"] = p

    # HEST directory metadata (via get_test_splits) → group by Sample_type/Location *if present*
    for group_col in ["Sample_type", "Location"]:
        if group_col in df_long.columns and df_long[group_col].notna().any():
            fig = plot_corrs_by_sample(df_long, dataset_name, encoder_name, group_by=group_col)
            p = outdir / f"per_sample_by_{group_col}.png"
            fig.savefig(p, dpi=200, bbox_inches="tight"); plt.close(fig)
            arts[f"per_sample_by_{group_col}"] = p

    if show:
        for k, v in arts.items():
            print(f"Saved {k}: {v}")

    return arts
```
